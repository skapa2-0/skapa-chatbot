```python

# Benchmark Ollama
# Ce script mesure le temps de réponse et le débit pour un modèle donné.

import os
import time
import statistics
import requests
from dotenv import load_dotenv

load_dotenv("../backend/.env", override=True)

MODEL = os.getenv("OLLAMA_MODEL", "llama3.1:8b")
OLLAMA_HOST = os.getenv("OLLAMA_HOST", "http://localhost:11434")
API_KEY = os.getenv("OLLAMA_API_KEY")
PROMPT = (
    "Explique en 5 phrases pourquoi les benchmarks sont utiles pour comparer "
    "des modèles de langage et comment mesurer la latence et le débit."
)
RUNS = 5

headers = {}
if OLLAMA_HOST.startswith("https://ollama.com"):
    if not API_KEY:
        raise ValueError("OLLAMA_API_KEY est requis pour l'API Ollama cloud")
    headers["Authorization"] = f"Bearer {API_KEY}"

# Vérifie que le service répond
try:
    r = requests.get(f"{OLLAMA_HOST}/api/tags", headers=headers, timeout=30)
    print("Status /api/tags:", r.status_code)
    print("Models disponibles:")
    print(r.json())
except Exception as e:
    print("Erreur de connexion à Ollama:", e)


def benchmark_once(prompt: str):
    start = time.perf_counter()
    response = requests.post(
        f"{OLLAMA_HOST}/api/generate",
        json={"model": MODEL, "prompt": prompt, "stream": False},
        headers=headers,
        timeout=300,
    )
    elapsed = time.perf_counter() - start

    try:
        body = response.json()
    except ValueError:
        body = {"response": response.text}

    answer = body.get("response", "")
    tokens = len(answer.split())
    tps = tokens / elapsed if elapsed > 0 else 0

    return {
        "status_code": response.status_code,
        "elapsed_sec": elapsed,
        "tokens": tokens,
        "tokens_per_sec": tps,
        "preview": answer[:120],
    }

results = [benchmark_once(PROMPT) for _ in range(RUNS)]

for i, r in enumerate(results, start=1):
    print(f"Run {i}: status={r['status_code']} | elapsed={r['elapsed_sec']:.2f}s | tokens={r['tokens']} | tokens/sec={r['tokens_per_sec']:.2f}")

latencies = [r["elapsed_sec"] for r in results]
tps = [r["tokens_per_sec"] for r in results]

print("\nMoyenne latence:", round(statistics.mean(latencies), 3), "s")
print("Médiane latence:", round(statistics.median(latencies), 3), "s")
print("P95 latence:", round(sorted(latencies)[min(len(latencies)-1, int(0.95 * len(latencies)) - 1)], 3), "s")
print("Moyenne tokens/sec:", round(statistics.mean(tps), 3))

# Exemple de sortie exploitable dans un DataFrame
import pandas as pd

df = pd.DataFrame(results)
print("\nDataFrame résultats:")
display(df)
```

## Benchmark multi-providers

Compare Ollama (local + cloud), Groq, Anthropic Claude et Gemini sur le même prompt.
Métriques : latence moyenne/médiane, tokens générés, tokens/sec.


```python
import os
import time
import statistics
import requests
import pandas as pd
from dotenv import load_dotenv

load_dotenv("../backend/.env", override=True)

PROMPT = (
    "Explique en 5 phrases pourquoi les benchmarks sont utiles pour comparer "
    "des modèles de langage et comment mesurer la latence et le débit."
)
RUNS = 5
TIMEOUT = 90

# (provider, model_id, label)
# Pour Ollama cloud : choisis un modèle visible dans la 1re cellule (/api/tags).
MODELS = [
    ("ollama_local", "llama3.2:1b",             "ollama-local llama3.2:1b"),
    ("ollama_cloud", "gpt-oss:20b",             "ollama-cloud gpt-oss:20b"),
    ("groq",         "llama-3.1-8b-instant",    "groq llama-3.1-8b-instant"),
    ("groq",         "llama-3.3-70b-versatile", "groq llama-3.3-70b-versatile"),
    ("gemini",       "gemini-2.0-flash",        "gemini-2.0-flash"),
]
```

```python
# ── Un appel non-streaming par provider — retourne le texte généré ──
def call_ollama_local(model, prompt):
    r = requests.post(
        "http://localhost:11434/api/chat",
        json={"model": model, "stream": False,
              "messages": [{"role": "user", "content": prompt}]},
        timeout=TIMEOUT,
    )
    r.raise_for_status()
    return r.json()["message"]["content"]


def call_ollama_cloud(model, prompt):
    key = os.getenv("OLLAMA_API_KEY")
    r = requests.post(
        "https://ollama.com/api/chat",
        headers={"Authorization": f"Bearer {key}"},
        json={"model": model, "stream": False,
              "messages": [{"role": "user", "content": prompt}]},
        timeout=TIMEOUT,
    )
    r.raise_for_status()
    return r.json()["message"]["content"]


def call_groq(model, prompt):
    key = os.getenv("GROQ_API_KEY")
    r = requests.post(
        "https://api.groq.com/openai/v1/chat/completions",
        headers={"Authorization": f"Bearer {key}"},
        json={"model": model,
              "messages": [{"role": "user", "content": prompt}]},
        timeout=TIMEOUT,
    )
    r.raise_for_status()
    return r.json()["choices"][0]["message"]["content"]


def call_anthropic(model, prompt):
    key = os.getenv("ANTHROPIC_API_KEY")
    r = requests.post(
        "https://api.anthropic.com/v1/messages",
        headers={"x-api-key": key, "anthropic-version": "2023-06-01",
                 "content-type": "application/json"},
        json={"model": model, "max_tokens": 1024,
              "messages": [{"role": "user", "content": prompt}]},
        timeout=TIMEOUT,
    )
    r.raise_for_status()
    return "".join(b.get("text", "") for b in r.json().get("content", []))


def call_gemini(model, prompt):
    key = os.getenv("GEMINI_API_KEY")
    r = requests.post(
        f"https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent?key={key}",
        json={"contents": [{"parts": [{"text": prompt}]}]},
        timeout=TIMEOUT,
    )
    r.raise_for_status()
    return r.json()["candidates"][0]["content"]["parts"][0]["text"]


CALLERS = {
    "ollama_local": call_ollama_local,
    "ollama_cloud": call_ollama_cloud,
    "groq":         call_groq,
    "anthropic":    call_anthropic,
    "gemini":       call_gemini,
}
```

```python
# ── Boucle de benchmark : RUNS appels, moyenne/médiane/tokens/sec ──
def benchmark(provider, model, prompt, runs=RUNS):
    caller = CALLERS[provider]
    latencies, token_counts = [], []
    error = None
    for _ in range(runs):
        try:
            t0 = time.perf_counter()
            text = caller(model, prompt)
            latencies.append(time.perf_counter() - t0)
            token_counts.append(len(text.split()))
        except Exception as e:
            error = str(e)[:200]
            break
    if error or not latencies:
        return {"provider": provider, "model": model, "error": error or "no runs"}
    avg_lat = statistics.mean(latencies)
    avg_tok = statistics.mean(token_counts)
    return {
        "provider":          provider,
        "model":             model,
        "avg_latency_s":     round(avg_lat, 3),
        "median_latency_s":  round(statistics.median(latencies), 3),
        "avg_tokens":        round(avg_tok, 1),
        "tokens_per_sec":    round(avg_tok / avg_lat, 2) if avg_lat else 0,
        "runs":              len(latencies),
    }


results = []
for provider, model, label in MODELS:
    print(f"→ {label} ...", end=" ", flush=True)
    r = benchmark(provider, model, PROMPT)
    results.append(r)
    if "error" in r:
        print(f"ERREUR : {r['error']}")
    else:
        print(f"{r['avg_latency_s']}s | {r['tokens_per_sec']} tok/s")

df = pd.DataFrame(results)
df_ok = df[~df.get("error", pd.Series(dtype=object)).notna()] if "error" in df.columns else df
df_sorted = df_ok.sort_values("tokens_per_sec", ascending=False).reset_index(drop=True)
display(df_sorted)
```

```python
df_sorted.to_csv("benchmark_results.csv", index=False)
print("Sauvegardé : benchmark_results.csv")
```
